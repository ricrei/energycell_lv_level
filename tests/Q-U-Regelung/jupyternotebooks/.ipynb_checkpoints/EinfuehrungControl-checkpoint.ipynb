Am Beispiel der Standardregelstrecke werden die wichtigsten Funktionen gezeigt

Zunächst sind drei Python-Pakete zu laden

```python
import numpy as np
from  control.matlab import *
import matplotlib.pyplot as plt
```

Das Numpy-Paket ist nicht immer erforderlich, da die Eingabe der Koeffizienten eines Polynoms auch als Liste erfolgen kann. 
Mit dem * werden alle Matlab-kompatiblen Funktionen der Control-Toolbox geladen, wobei die Funktionen direkt verwendet werden können. Mit tf() wird z.B. eine Übertragungsfunktion erzeugt, die mit anderen verknüpft werden kann.

Die Eingabe der Streckenparameter als Übertragungsfunktion $G(s)$, Ausgabe der Strecke und Berechnung der Zeitkonstanten:


```python
b0 = 1.
a0 = 1. ; a1 = 10.; a2 = 31. ; a3 = 30. 
num  = np.array([b0])
den  = np.array([a3, a2, a1, a0])
strecke = tf(num,den)
print('G(s) = ',strecke)
print('Zeitkonstanten = ',-1/pole(strecke))
```

Die Pole bzw. Eigenwerte eines Systems sind mit der Funktion pole() berechenbar. Das gilt auch für eine Beschreibung im Zustandsraum.

Die Pole und Nullstellen können mit der Funktion pzmap() berechnet und graphisch dargestellt werden.

Die Reihenschaltung mit einem P-Regler und Berechnung des Bode-Diagramms des offenen Kreises erfolgt gemäß:

```python
regler = 1.0
G0 = regler*strecke

bb = bode(G0)
```

Ein PID-Regler wird nachfolgend in Form einer Übertragungsfunktion dargestellt, wobei die Parameter

```python
kr =2.0; tn = 8.0
tv = 1.875 ; t1 = tv/10.0
```

gemäß Kompensationsentwurf ermittelt wurden. Berechnung des offenen Kreises durch Multiplikation der Übertragungsfunktionen von Regler und Strecke und Ausgabe des Reglers:

```python
regler = tf([kr*(tv+t1)*tn, kr*(tn+t1), kr],[tn*t1, tn, 0])
G0 = regler*strecke
print ('R(s) = ',regler)
```

Der Frequenzgang des offenen Kreises lässt sich auch in Form einer Ortskurve ausgeben. Dazu gibt es die Funktion nyquist(), die auch den Plot um die negativen Frequenzen erweitert. Um dies zu vermeiden ist folgender code erforderlich:

```python
real, imag, ww = nyquist(G0,Plot=False)
plt.plot(real, imag, "b")

plt.title('Ortskurve')
plt.grid()
```

Die Wurzelortskurve des offenen Kreises wird mit dem folgendem Befehl berechnet.

```python
rr = rlocus(G0)
```

Durch das Anklicken der Wurzelortsurve werden die dazugehörige Verstärkung und Dämpfung ausgegeben. Falls keine Variable für die Rückgabe der berechneten Werte angegeben wird, in diesem Falle rr, dann erfolgt die Ausgabe in der Konsole.  Das gilt auch für bode().

Ein weiteres Werkzeug für die Analyse und Synthese von Regelkreisen stellt das sisotool() dar, siehe Doku.

Der Regelkreis kann mit Hilfe der feedback() Funktion berechnet werden. Die Simulation der Sprungantwort des Kreises benötigt die Funktion step():

```python
regelkreis = feedback(G0,1)
x, t = step(regelkreis)

xx, t = step(strecke)
```

Plot Sprungantwort Strecke und Regelkreis

```python
plt.plot(t, x, t, xx) 
plt.grid()
plt.title('Sprungantwort Strecke und Regelkreis')
plt.xlabel('t [s]')
```
